```python
#cafe bills
menu = {"burger": 150,"pizza": 300,"pasta": 220,"sandwich": 120}
item=input("enter the item: ")
quantity=int(input("enter the quantity: "))
if item in menu:
    bill=menu[item]*quantity
    if bill>500:
        discount=10
    elif bill>1000:
        discount=15
    else:
        discount=0
    discount_amount=bill*discount/100
    total_amount=bill-discount_amount
    print("total amount= ",total_amount)
else:
    print("item is not available")




```

```python
#stock calculation
stock = {"laptop": 5,"mouse": 0,"keyboard": 8,"monitor": 2}
product=input("enter the product: ")
if product not in stock:
    print("product not found")
elif stock[product]==0:
    print("out of stock")
elif stock[product]>=1 and stock[product]<=3 :
    print("low stock")
else:
    print("stock is available")
```
