```python
#Take a number and check whether it is an **Armstrong number**. Example: `153 → Armstrong number`
num=input("enter the number:")
sum=0
count=len(num)
for i in num:
    i=int(i)
    value=i**count
    sum=sum+value
if int(num)==sum:
    print(num,"is an armstrong num")
else:
    print(num,"is not an armstrong number")
```

```python
#Create a simple **menu-driven program*** `1` → Add two numbers* `2` → Subtract two numbers* `3` → Multiply two numbers* `4` → Exit Keep displaying the menu until the user chooses 4.
while True:
    print("\n 1.Addition ,2.Substraction ,3.Multiplication ,4.Exit")
    choice=int(input("enter the number between 1-4"))
    if choice==4:
        break
    num1=int(input("enter the first number:"))
    num2=int(input("enter the second number:"))
    print("Exit")
    if choice==1:
        print("sum of two numbers is",num1+num2)
    elif choice==2:
        print("difference between two number is ",num1-num2)
    else:
        print("product of two numbers is",num1*num2)
```

```python
#Create a simple **ATM PIN checker**.Give the user a maximum of **3 attempts** to enter the correct PIN. Display `"Account Locked"` after 3 incorrect attempts.
pin=1234
attempt=0
while True:
    pin_new=int(input("enter the PIN"))
    if pin_new!=pin:
        attempt=attempt+1
        print("incorrect PIN")
       
        if  attempt==3:
            print("Account Locked")
            break
    else:
        print("your transaction is in process")
        break
            
```
