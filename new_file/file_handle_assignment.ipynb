```python
#Q6. Count the vowels, consonants, digits and spaces in a file. Print the counts as a dictionary.
with open("profile.txt","r") as file:
    text=file.read()
    vowels=0
    consonants=0
    digits=0
    spaces=0
    for ch in text:
        if ch in "aeiou":
            vowels+=1
        elif ch.isalpha():
            consonants+=1
        elif ch.isdigit():
            digits+=1
        elif ch==" ":
            spaces+=1
    result={
        "vowels":vowels,
        "consonants":consonants,
        "digits":digits,
        "spaces":spaces
        }
    print(result)


```

```python
#Q10. Write a function replace_word(filename, old, new) that saves the changed text to a new file and returns how many replacements it made.
def replace_word(filename,old,new):
    with open(filename,"r") as file:
        content=file.read()
        count=content.count(old)
        new_content=content.replace(old,new)
    with open("new_file","w") as file:
        file.write(new_content)
    return f"Total replacements:{count}"
replace_word("para.txt","all","not")
```

```python
def check_balance():
    balance=0
    with open("transactions.txt","r") as file:
        lines=file.readlines()
        for i in lines:
            kind,amount=i.split(",")
            if kind=="deposit":
                balance+=int(amount)
            elif kind=="withdraw":
                balance-=int(amount)
    return balance
check_balance()
def add_transaction(kind,amount):
    pay=check_balance()
    if kind=="withdraw":
        if pay<=amount:
            print("invalid transaction")
        else:
            with open("transaction.txt","a") as file:
                file.write(f"{kind} , {amount}\n")
    else:
        with open("transactions.txt","a") as file:
            file.write(f"{kind} , {amount}\n")
    return pay
add_transaction("withdraw",20000)
```
