## String Problems


```python
TEXT="Aysha"
print(("First letter is:",TEXT[0]))
print("Last letter is:",TEXT[-1])
```

```python
#reversing using slice
TEXT="Nasni"
print(TEXT)
print("reverse of the word is:",TEXT[ : :-1])
```

```python
#Counting the number
TEXT="banana"
print(TEXT)
print(TEXT.count("a"))#count of letter a in this word is 3
```

```python
#Replace "JAVA" with "Python" from "JAVA is simple & powerfull"
Text="JAVA is simple & powerful"

print(Text.replace("JAVA" , "Python"))
```

```python
#Convert a string to uppercase and lowercase
TEXT="Vadakara"
print(TEXT.upper())
print(TEXT.lower())
TEXT="Aysha Nasni"
print(TEXT.capitalize())
print(TEXT.title())
```

```python
#splitting the string
TEXT="Python,Java,SQL"
print(TEXT.split(","))
```

```python
#joining the string
TEXT=["Python","Java","SQL"]
print("-".join(TEXT))
```

```python
#checking the availability
TEXT="Python"
print(TEXT)
print("thon" in TEXT)
```

```python
#printing character
TEXT="Programming"
print("every second character in this is:",TEXT[ 1: :2])
```
