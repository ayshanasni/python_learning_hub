```python
# Practical 1: Simple calculator that keeps running until user chooses to exit
while True:
    choice=int(input("Choose an option (1-5)"))
    if choice==5:
        print("Exit")
        break
    num1=int(input("enter the first number: "))
    num2=int(input("enter the second number: "))
    if choice==1:
        print(num1,"+",num2,"=",num1+num2)
    elif choice==2:
        print(num1 ,"-",num2,"=",num1-num2)
    elif choice==3:
        print(num1,"*",num2,"=",num1*num2)
    else:
        if num2!=0:
            print("result",num1/num2)
        else:
            print("error:division by zero")
else:
    print("invalid number-choose a number between 1-5")
    
```
