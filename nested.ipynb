```python
#Check whether a given number is an Armstrong number (e.g., 153 = 13 + 53 + 33).
num=int(input("enter the number"))
count=len(str(num))
sum=0
for i in str(num):
    sum=sum+int(i)**count
if num==sum:
    print("it is an armstrong number")
else:
    print("not an armstrong number")
    
```

```python
#Take a number, keep adding its digits together, and repeat the process until the result is a single digit (e.g., 9875 → 29 → 11 → 2).
number=input("enter the number")
while len(number)>1:
    sum=0
    for digit in number:
        sum=sum+int(digit)
    number=str(sum)
print(number)

```
